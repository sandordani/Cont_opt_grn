```python
import zipfile
import os
import numpy as np
import sys

sys.path.append('/home/sandor_daniel/work/2023-11-08_cont_opt_grn/notears')

from notears import linear, utils
import numpy as np
from tqdm import tqdm
from collections import defaultdict
import os

utils.set_random_seed(1)

with zipfile.ZipFile("/home/sandor_daniel/work/2023-11-08_cont_opt_grn/DREAM4_all_files.zip", 'r') as zip_ref:
    zip_ref.extractall('DREAM4')

for file in os.listdir('DREAM4'):
    if file.endswith(".zip"):
        with zipfile.ZipFile(os.path.join('DREAM4', file), 'r') as zip_ref:
            zip_ref.extractall('DREAM4')
        os.remove(os.path.join('DREAM4', file))

def load_data(directory, condition):
    data = [np.loadtxt(f'{directory}/{folder}/{folder}_timeseries.tsv', skiprows=1)[:,1:] for folder in sorted(os.listdir(directory)) if condition(folder)]
    return data

def estimate_dag(data, save_path):
    est = []
    for (i, x) in enumerate(data):
        print(i)
        W_notears = linear.notears_linear(x, lambda1=0, loss_type='l2')
        assert utils.is_dag(W_notears)
        est.append(W_notears)

    for (i, g) in enumerate(est):
        np.savetxt(f'{save_path}/dag_{i}.txt', g, delimiter=',')

    return est
```

```python


dream4_data_10 = load_data('DREAM4', lambda folder: 'size10_' in folder)
dream4_data_100 = load_data('DREAM4', lambda folder: 'size100_' in folder and 'multifactorial' not in folder)


dream4_est_10 = estimate_dag(dream4_data_10, 'out/dream4_10/notears')
dream4_est_100 = estimate_dag(dream4_data_100, 'out/dream4_100/notears')


```
